```python
import numpy as np
import time
from keras.datasets import mnist
from keras.utils import to_categorical
from keras.models import Sequential
from keras.layers import Conv2D, MaxPool2D, Flatten, Dense, Dropout, BatchNormalization
from sklearn.metrics import precision_score, recall_score
from keras.callbacks import EarlyStopping

# Load dataset
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Reshape the dataset
X_train = X_train.reshape(60000, 28, 28, 1).astype('float32') / 255
X_test = X_test.reshape(10000, 28, 28, 1).astype('float32') / 255

# One-hot encoding of labels
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# Create an enhanced custom model
custom_model_enhanced = Sequential()

# First convolutional block
custom_model_enhanced.add(Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=(28, 28, 1)))
custom_model_enhanced.add(BatchNormalization())
custom_model_enhanced.add(MaxPool2D(pool_size=(2, 2)))
custom_model_enhanced.add(Dropout(0.25))

# Second convolutional block
custom_model_enhanced.add(Conv2D(64, kernel_size=(3, 3), activation='relu'))
custom_model_enhanced.add(BatchNormalization())
custom_model_enhanced.add(MaxPool2D(pool_size=(2, 2)))
custom_model_enhanced.add(Dropout(0.25))

# Flatten and Dense layers
custom_model_enhanced.add(Flatten())
custom_model_enhanced.add(Dense(128, activation='relu'))
custom_model_enhanced.add(Dropout(0.5))
custom_model_enhanced.add(Dense(10, activation='softmax'))

# Compile the model
custom_model_enhanced.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Early stopping to prevent overfitting
early_stopping = EarlyStopping(monitor='val_loss', patience=2)

# Train the model and measure time
start_time = time.time()
custom_model_enhanced.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=10, batch_size=64, callbacks=[early_stopping])
training_time = time.time() - start_time

# Evaluate the model
loss, accuracy = custom_model_enhanced.evaluate(X_test, y_test)
print(f"Enhanced Custom Model Accuracy: {accuracy * 100:.2f}%")
print(f"Training Time: {training_time:.2f} seconds")

# Predictions for precision and recall
y_pred = custom_model_enhanced.predict(X_test)
y_pred_classes = y_pred.argmax(axis=1)
y_true_classes = y_test.argmax(axis=1)

precision = precision_score(y_true_classes, y_pred_classes, average='weighted')
recall = recall_score(y_true_classes, y_pred_classes, average='weighted')

print(f"Enhanced Custom Model Precision: {precision:.2f}")
print(f"Enhanced Custom Model Recall: {recall:.2f}")
```
